## Q1. Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.).
1. Convert text to lowercase and remove punctuation using re.
2. Tokenize the text into words and sentences.
3. Split using split() and word_tokenize() and compare how Python split and NLTK’s
word_tokenize() differ.
4. Remove stopwords (using NLTK's stopwords list).
5. Display word frequency distribution (excluding stopwords).

```python
# Q1: Text Preprocessing and Tokenization

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist


# Unique paragraph
text = """
Artificial intelligence is one of my favorite topics because it is changing the way we learn and work.
AI can analyze large amounts of data and help people solve complex problems.
I enjoy learning about machine learning because it combines programming, mathematics, and creativity.
AI-powered tools can also make everyday tasks faster and more convenient.
However, I believe that AI should always be used responsibly and ethically.
The future of artificial intelligence is exciting, and I would love to explore it further.
"""

print("Original Text:")
print(text)

# 1. Convert text to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("\nText after lowercase and punctuation removal:")
print(clean_text)

# 2. Tokenize text into words and sentences
word_tokens = word_tokenize(clean_text)
sentence_tokens = sent_tokenize(text)

print("\nWord Tokens:")
print(word_tokens)

print("\nSentence Tokens:")
print(sentence_tokens)

# 3. Compare split() and word_tokenize()
split_tokens = clean_text.split()

print("\nTokens using split():")
print(split_tokens)

print("\nTokens using word_tokenize():")
print(word_tokens)

print("\nNumber of tokens using split():", len(split_tokens))
print("Number of tokens using word_tokenize():", len(word_tokens))

print("\nComparison:")
print("split() simply separates text based on whitespace.")
print("word_tokenize() uses NLTK's tokenizer and handles linguistic tokenization more effectively.")

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in word_tokens
    if word not in stop_words
]

print("\nWords after removing stopwords:")
print(filtered_words)

# 5. Display word frequency distribution
freq_dist = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
for word, frequency in freq_dist.items():
    print(word, ":", frequency)
```

## Q2. Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall()
2. Remove stop words using NLTK’s stopword list
3. Perform stemming with PorterStemmer
4. Perform lemmatization with WordNetLemmatizer
5. Compare the stemmed and lemmatized outputs and explain when you’d prefer one over the other.

```python
# Q2: Stemming and Lemmatization

import re
import nltk
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer



# Same paragraph from Q1
text = """
Artificial intelligence is one of my favorite topics because it is changing the way we learn and work.
AI can analyze large amounts of data and help people solve complex problems.
I enjoy learning about machine learning because it combines programming, mathematics, and creativity.
AI-powered tools can also make everyday tasks faster and more convenient.
However, I believe that AI should always be used responsibly and ethically.
The future of artificial intelligence is exciting, and I would love to explore it further.
"""

# 1. Extract words containing only alphabets
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Words extracted using re.findall():")
print(words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("\nWords after removing stopwords:")
print(filtered_words)

# 3. Stemming using PorterStemmer
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Lemmatization using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Compare outputs
print("\nComparison of Stemming and Lemmatization:")
print("Stemming:", stemmed_words)
print("Lemmatization:", lemmatized_words)

print("\nExplanation:")
print("Stemming removes word endings to produce a root-like form.")
print("It is faster but may produce words that are not actual English words.")

print("\nLemmatization considers the meaning and vocabulary of words.")
print("It usually produces valid dictionary words and is preferred when linguistic accuracy is important.")

print("\nPreference:")
print("I would prefer stemming when speed and simplicity are more important.")
print("I would prefer lemmatization when the meaning and grammatical correctness of words are important.")
```

## Q3. Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representation.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.


```python
# Q3: Bag of Words and TF-IDF

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Three short texts
texts = [
    "Artificial intelligence is transforming modern technology and improving productivity.",
    "Machine learning helps computers learn patterns from data and make predictions.",
    "Cloud computing provides scalable storage and powerful computing resources."
]

# 1. Bag of Words using CountVectorizer
count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Representation:")
print(bow_matrix.toarray())

# 2. TF-IDF using TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

# 3. Top 3 keywords from each text
print("\nTop 3 Keywords from Each Text:")
print("--------------------------------")

for i, text in enumerate(texts):
    scores = tfidf_matrix[i].toarray().flatten()
    
    top_indices = scores.argsort()[-3:][::-1]
    
    print(f"\nText {i+1}:")
    print(text)
    
    print("Top 3 Keywords:")
    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 4))
```

## Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:
a. Jaccard Similarity using sets
b. Cosine Similarity using TfidfVectorizer + cosine_similarity()
c. Analyze which similarity metric gives better insights in your case.


```python
# Q4: Jaccard Similarity and Cosine Similarity

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Two texts about different technologies

text1 = """
Artificial intelligence enables computers to perform tasks that normally require human intelligence.
AI systems can learn from data and recognize patterns.
Machine learning is an important part of artificial intelligence.
AI is widely used in healthcare, education, and automation.
It helps organizations make better decisions and improve productivity.
"""

text2 = """
Blockchain is a distributed technology used to store data securely.
It records transactions in a decentralized and transparent manner.
Blockchain uses cryptography to protect stored information.
It is widely used in cryptocurrencies and digital transactions.
The technology can also be applied to supply chain management.
"""

# Preprocessing function
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = text.split()
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Tokens from Text 1:")
print(tokens1)

print("\nTokens from Text 2:")
print(tokens2)

# 1. Jaccard Similarity
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard_similarity = len(intersection) / len(union)

print("\nJaccard Similarity:")
print(round(jaccard_similarity, 4))

# 2. Cosine Similarity using TF-IDF
texts = [text1, text2]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(texts)

cosine_sim = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("\nCosine Similarity:")
print(round(cosine_sim[0][0], 4))

# 3. Analysis
print("\nAnalysis:")
print("Jaccard Similarity compares the overlap between the unique words in the two texts.")
print("Cosine Similarity considers the TF-IDF importance of the words.")
print("For these texts, Cosine Similarity generally gives better insights because")
print("it considers word importance instead of treating every word equally.")
```
