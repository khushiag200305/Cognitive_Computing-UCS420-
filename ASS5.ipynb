### Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:

### a) Addition of 2 in all the element
### b) Multiply 3 with all the elements
### c) Divide every element by 2

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])

print("Original Array:", arr)

print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

### Q.2 Questions on Basic NumPy Array:

### a) Reverse the NumPy array:
### arr = np.array([1, 2, 3, 6, 4, 5])
### b) Find the most frequent value and their indice(s) in the following arrays:
### i. x = np.array([1,2,3,4,5,1,2,1,1,1])
### ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

```python
# part a)
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original Array:", arr)
print("Reversed Array:", arr[::-1])
```

```python
# part b)
x = np.array([1,2,3,4,5,1,2,1,1,1])
freq = np.bincount(x)
most_freq = np.argmax(freq)
indices = np.where(x == most_freq)[0]
print("Most Frequent Value:", most_freq)
print("Indices:", indices)
```

```python
y = np.array([1,1,1,2,3,4,2,4,3,3])
unique, counts = np.unique(y, return_counts=True)
max_count = np.max(counts)
most_frequent_values = unique[counts == max_count]
print("Most Frequent Value(s):", most_frequent_values)
for value in most_frequent_values:
    indices = np.where(y == value)[0]
    print(f"Value {value} occurs at indices:", indices)
```

### Q.4 Write program to create an 1-D NumPy array named <<"Your Name">> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?


```python
khushi = np.linspace(10, 100, 25)
print("Array:", khushi)
print("Dimensions:", khushi.ndim)
print("Shape:", khushi.shape)
print("Total Elements:", khushi.size)
print("Data Type:", khushi.dtype)
print("Bytes Consumed:", khushi.nbytes)
reshaped = khushi.reshape(5,5)
print("\nReshaped Array:")
print(reshaped)
print("\nTranspose:")
print(reshaped.T)
print("\nTranspose of Original 1-D Array:")
print(khushi.T)
```

### Q5. Create a 2-D Array of three rows and four columns, named ucs420_<>your_name<> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<>your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<>your_name>>.

```python
ucs420_khushi = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print("Mean:", np.mean(ucs420_khushi))
print("Median:", np.median(ucs420_khushi))
print("Maximum:", np.max(ucs420_khushi))
print("Minimum:", np.min(ucs420_khushi))
print("Unique Elements:", np.unique(ucs420_khushi))
reshaped_ucs420_khushi = ucs420_khushi.reshape(4,3)
print("\nReshaped Array:")
print(reshaped_ucs420_khushi)
resized_ucs420_khushi = np.resize(ucs420_khushi,(2,3))
print("\nResized Array:")
print(resized_ucs420_khushi)
```
