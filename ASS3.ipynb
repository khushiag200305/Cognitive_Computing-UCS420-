# Q6. Create a sample dataset(employee.csv)containing information about employees in a company 

```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)

employees
```

### a) Shape of the DataFrame



```python
print("Shape of DataFrame:", employees.shape)
```

### b) Summary of the DataFrame that includes the data types and non-null count for each column 

```python
employees.info()
```

```python
employees.describe()
```

### d) Display the first 5 rows and last 3 rows of the dataset.

```python
print("First 5 rows:")
print(employees.head(5))

print("\nLast 3 rows:")
print(employees.tail(3))
```

### e) Calculate the following statistics from the dataset:
### i. The average salary of employees.
### ii. The total bonus paid to all employees.
### iii. The youngest employee's age.
### iv. The highest performance rating.

```python
print("Average salary:", employees["Salary"].mean())

print("Total bonus:", employees["Bonus"].sum())

print("Youngest employee's age:", employees["Age"].min())

print("Highest performance rating:", employees["Rating"].max())
```

### f) Sort the DataFrame by the Salary column in descending order.

```python
employees.sort_values(by="Salary", ascending=False)
```

### g) Add a new column that categorizes employees based on their performance rating:
### i. Excellent for ratings >= 4.5
### ii. Good for ratings >= 4.0 but < 4.5
### iii. Average for ratings < 4.0

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(categorize_rating)

employees
```

### h) Identify missing values in the DataFrame.

```python
employees.isnull().sum()
```

### i) Rename the Employee_ID column to ID.

```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)

employees
```

### j) Find all employees who:
### i. Have more than 5 years of experience.
### ii. Belong to the IT department.

```python
print("Employees with more than 5 years of experience:")

employees[employees["Years_of_Experience"] > 5]
```

```python
print("Employees belonging to the IT department:")

employees[employees["Department"] == "IT"]
```

### k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.

```python
employees["Tax"] = employees["Salary"] * 0.10

employees
```

### l) Save the modified DataFrame (with added columns) to a new CSV file.

```python
employees.to_csv("employees_modified.csv", index=False)

print("Modified DataFrame saved successfully.")
```
